```python
import os
import shutil
import random
```

```python
#set random seed for reproducibility
random.seed(42)
```

```python
#original dataset paths
source_dir = "classes"
bareland_dir = os.path.join(source_dir, "Bareland")
beach_dir = os.path.join(source_dir, "Beach")
dense_residential_dir = os.path.join(source_dir, "Dense-Residential")
desert_dir = os.path.join(source_dir, "Desert")
farmland_dir = os.path.join(source_dir, "Farmland")
forest_dir = os.path.join(source_dir, "Forest")
illegal_mining_dir = os.path.join(source_dir, "Illegal-Mining")
legal_mining_dir = os.path.join(source_dir, "Legal-Mining")
mountain_dir = os.path.join(source_dir, "Mountain")
sparse_residential_dir = os.path.join(source_dir, "Sparse-Residential")
```

```python
#new structured dataset paths
target_base = "data"
splits = ['train', 'val']
categories = {
    'Bareland': 'Bareland', 
    'Beach': 'Beach', 
    'Dense-Residential':'Dense-Residential', 
    'Desert':'Desert', 
    'Farmland':'Farmland', 
    'Forest':'Forest', 
    'Illegal-Mining':'Illegal-Mining', 
    'Legal-Mining':'Legal-Mining', 
    'Mountain':'Mountain', 
    'Sparse-Residential':'Sparse-Residential'
}
split_ratio = 0.8  #80% train, 20% val
```

```python
#create target dirs
for split in splits:
    for cat in categories.values():
        os.makedirs(os.path.join(target_base, split, cat), exist_ok=True)
```

```python
#function to split and move
def split_and_copy(src_folder, label_name):
    images = [f for f in os.listdir(src_folder) if f.lower().endswith(('.jpg', '.png'))]
    random.shuffle(images)

    split_index = int(len(images) * split_ratio)
    train_imgs = images[:split_index]
    val_imgs = images[split_index:]

    for img in train_imgs:
        shutil.copy(
            os.path.join(src_folder, img),
            os.path.join(target_base, 'train', label_name, img)
        )
    for img in val_imgs:
        shutil.copy(
            os.path.join(src_folder, img),
            os.path.join(target_base, 'val', label_name, img)
        )
```

```python
split_and_copy(bareland_dir, 'Bareland')
split_and_copy(beach_dir, 'Beach')
split_and_copy(dense_residential_dir, 'Dense-Residential')
split_and_copy(desert_dir, 'Desert')
split_and_copy(farmland_dir, 'Farmland')
split_and_copy(forest_dir, 'Forest')
split_and_copy(illegal_mining_dir, 'Illegal-Mining')
split_and_copy(legal_mining_dir, 'Legal-Mining')
split_and_copy(mountain_dir, 'Mountain')
split_and_copy(sparse_residential_dir, 'Sparse-Residential')
```
